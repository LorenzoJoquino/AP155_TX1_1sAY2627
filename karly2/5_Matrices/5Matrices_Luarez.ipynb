$$\begin{align} T_1\cos(\alpha) + T_2\cos(\beta) = 0 \tag{1}\\ 
T_1\sin(\alpha) + T_2\sin(\beta) = Mg \tag{2} \end{align}$$
Where Eq. (1) are EoM along $x$ and Eq. (2) describes EoM along $y$.

2. For both problems, the function used to solve the system of linear equations is np.linalg.solve(). The function works by calculating the exact solution array $\mathbf x$ for the matrix equation $\mathbf A \mathbf b = \mathbf x$ where $\mathbf A$ is square and full rank, i.e. its rows and columns are linearly independent. Its inputs are an an array-like coefficient matrix $A$ and the output array $b$ and its output is the solution array $x$ with identical shape to $b$. The solutions are computed using the LAPACK routine _gesv. The dgesv routine solves real systems $AX=B$ where $A$ is $N \times N$ and $B$ is $N x$ the number of right hand  sides using LU factorization this performs the decomposition $A=PLU$ where $L$ and $U$ are lower and upper triangular matrices and $P$ is the permutation matrix [1]. The factored form is then used to solve an equivalent matrix equation
$$L(Ux)=Pb$$
where $y=Ux$ is unknown. Since L is lower triangular, then the variables $y_i$ can be known immediately. Then $x$ is computed using the upper triangular using the equation
$$Ux=y$$

4. Now repeat your calculation for the case where there are $N=10\,000$ internal junctions.  This part is not possible using standard tools like the *solve* function.  You need to make use of the fact that the matrix~$\vec{A}$ is banded.  

For $N=6$ junctions, we have a system of six linear equations:
$$3V_1-V_2-V_3+0+0+0=5V$$

$$-V_1+4V_2-V_3-V_4+0+0=5V$$

$$-V_1-V_2+4V_3-V_4-V_5+0=0$$

$$0-V_2-V_3+4V_4-V_5-V_6=0$$

$$0+0-V_3-V_4+4V_5-V_6=0$$

$$0+0+0-V_4-V_5+3V_6=0$$

Rewriting the system of equations in the form $\mathbf{A}\mathbf{\vec{b}}=\mathbf{\vec{w}}$:

$$\begin{bmatrix}
3 & -1 & -1 & 0 & 0 & 0\\
-1 & 4 & -1 & -1 & 0 & 0\\
-1 & -1 & 4 & -1 & -1 & 0\\
0 & -1 & -1 & 4 & -1 & -1\\
0 & 0 & -1 & -1 & 4 & -1\\
0 & 0 & 0 & -1 & -1 & 3\\
\end{bmatrix} 
\begin{bmatrix} V_1\\V_2\\V_3\\V_4\\V_5\\V_6\end{bmatrix} 
= \begin{bmatrix}5\\5\\0\\0\\0\\0\end{bmatrix}$$ 

```python
# Using np.linalg.solve() to find V_is:

A = np.array([[3,-1,-1,0,0,0],[-1,4,-1,-1,0,0],[-1,-1,4,-1,-1,0],[0,-1,-1,4,-1,-1],[0,0,-1,-1,4,-1],[0,0,0,-1,-1,3]])
w = np.array([5,5,0,0,0,0])

Vs = np.linalg.solve(A,w)

for i, Vi in enumerate (Vs, start = 1):
  print(f"V_{i} = {Vi}")
```

In general, to solve for a system with $N$ equations, we use the fact that the coefficient matrix $\mathbf A$ is a banded matrix. We recreate the $N \times N$ matrix $\mathbf A$ for $N$ equations and use np.linalg.solve() to find solutions to that system.

```python
N = 10000

def solve_banded(N):
  # We create a 5xN array since A has 5 non-zero diagonals
  diag_A = np.zeros((5, N)) 
  diag_A[0,2:] = -1 # 2nd UD (upper diagonal)
  diag_A[1,1:] = -1 # 1st UD
  diag_A[2,:] = 4 # Main Diagonal
  diag_A[3, :-1] = -1 # 1st LD (lower diagonal)
  diag_A[4, :-2] = -1 # 2nd LD
  # The (0,0) and (N,N) element:
  diag_A[2,0] = 3
  diag_A[2,-1] = 3
  # creating the output vector w
  w = np.zeros(N)
  w[0] = 5
  w[1] = 5
  
  A = np.zeros((N,N)) # create an NxN storage
  for j in range(N): # iterate each column with index j
    l = 2 # no. of lower diagonal bands
    u = 2 # no. of upper diagonal bands
    imin = max(0, j - u) # row elements above the main diagonal
    imax = min(N, j + l + 1) # row elements below the main diagonal
    for i in range(imin, imax): #iterate over each column with index j
        A[i, j] = diag_A[u + i - j, j] # copy elements from the diag_A into the nonzero diagonals of the coefficient matrix A
            
  Vs = np.linalg.solve(A, w)
  return Vs, A

# For N = 1000
Vs, A = solve_banded(N)
print("Coefficient matrix A:")
print(A)
print("Vi:")
print(Vs)
```

### Some codes for solving matrix problems

* solving matrix equations ($\bf{A}\vec{x} = \vec{b}$) -> `np.linalg.solve`
* LU-decomposition ($\bf{A} = \bf{L}\bf{U}$) -> `scipy.linalg.lu`
* matrix inversion ($\bf{A}^{-1}$) -> `np.linalg.inv`
* QR-decomposition ($\bf{A} = \bf{Q}\bf{R}$) -> `scipy.linalg.qr`
* eigenvalue problem ($\bf{A}\vec{x} = \lambda \vec{x}$) -> `np.linalg.eig`

Solving matrix problems by themselves isn't difficult. There is an abundance of linear algebra libraries that are ready to use: from the battle-tested ones (the classic [BLAS](https://www.netlib.org/blas/) and [LAPACK](https://www.netlib.org/lapack/)), HPC tools for large problems ([ScaLAPACK](https://netlib.org/scalapack/)), to hardware-accelerated ones often with GPUs ([cuBLAS](https://docs.nvidia.com/cuda/cublas/index.html) and [cuSolver](https://docs.nvidia.com/cuda/cusolver/index.html) for NVIDIA).
