### Problem 1: 2D Wave interference (Gezerlis Problem 1.15)

Assume that the combined eﬀect of two objects dropped in a water basin can be modelled by:
$$ W(\textbf{r}) = A \sin (k|\textbf{r}-\textbf{r}_0|) + A  \sin (k|\textbf{r}-\textbf{r}_1|). $$ You can think of this as an updated version of Eq. (1.3) in the book, where we are dealing with the linear addition of sinusoidal waves, instead of electric-field contributions. For simplicity, take $A=1$, $k=2\pi/0.3$, $r_0 = (−1, 0)$ , and $r_1 = (1, 0)$. Since we are now faced with a density plot, you should employ, NOT `streamplot()` but, `imshow()` from `Matplotlib`; in each of $x$ and $y$ your plot should extend from $−4$ to $+4$.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt

def makewave(xs, ys, sourcetopos):
    A = 1.0
    k = 2 * np.pi / 0.3
    n = len(xs)
    
    #reference to visualizing electric fields notes
    xis = [[0. for _ in range(n)] for _ in range(n)]
    
    for j, x in enumerate(xs):
        for k_idx, y in enumerate(ys):
            for source, pos in sourcetopos.items():
                posx, posy = pos
                R = sqrt((x - posx)**2 + (y - posy)**2)
                xis[k_idx][j] += A * np.sin(k * R)
                
    return xis


def plotwave(boxl, n):
    #grid points
    xs = [-boxl + i*2*boxl/(n-1) for i in range(n)]
    ys = xs[:]
    
    #source configs
    sources_left = {1: (-1, 0), 2: (1, 0)}
    sources_right = {1: (-1, 0), 2: (1, 0), 3: (0, 1), 4: (0, -1)} 
    
    #wavefields for left&right box
    xis_left = makewave(xs, ys, sources_left)
    xis_right = makewave(xs, ys, sources_right)
    
    #converting to np
    xis_left = np.array(xis_left)
    xis_right = np.array(xis_right)
    
    #side-by-side
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
    
    #plot left box
    ax1.imshow(xis_left, extent=[-boxl, boxl, -boxl, boxl], cmap='Blues', origin='lower')
    
    #plot right box
    ax2.imshow(xis_right, extent=[-boxl, boxl, -boxl, boxl], cmap='Blues', origin='lower')
    
    fig.savefig("interference_pattern.png")
    plt.show()

plotwave(4.0, 300)
```

#### Problem 1 code summary

This script illustrates how waves from different sources overlap and interact with one another. It creates a grid of x and y coordinates, then uses nested loops to calculate the wave height at every point by adding up the sine waves from each source. I wrote this code this way to understand and apply the code used in the "visualizing electric fields" part in the given notes, while also using imshow() from Matplotlib to match the pictures given in the instructions.

### Problem 2: Listcomps-maxxing

Import the `random` library (docs: https://docs.python.org/3/library/random.html). We will be using the `random.randint` function to pick random integers.

a. Create a function `random_integer_matrix(nrows, ncols)` that generates a matrix in the form of "list-of-lists" whose elements are random integers from $1$ to $9$. Each "sub-list" inside the main list is a row in the matrix.

```python
#generate a matrix filled with random numbers from 1 to 9
def random_integer_matrix(nrows, ncols):
    #outer makes rows; inner fills each row with numbers
    return [[random.randint(1, 9) for _ in range(ncols)] for _ in range(nrows)]
```

b. Create a function `scalar_product(list1, list2)` where you multiply each element at the same index in the two lists and sum/accumulate the results.

```python
#dot product
def scalar_product(list1, list2):
    #zip() to pair numbers together; sum() to add multiplied pairs
    return sum(a * b for a, b in zip(list1, list2))
```

c. Create a function `transpose_matrix(orig_matrix)` which returns the transpose of any matrix also in "list-of-lists" form. Do not assume the size of the matrix; figure it out within the function. Double-check that your implementations above work. Create a $3\times4$ random integer matrix, print it out and its transpose. The values and shape between the two should be consistent.

```python
def transpose_matrix(orig_matrix):

    nrows = len(orig_matrix)
    ncols = len(orig_matrix[0])
    # swap row r and column c
    return [[orig_matrix[r][c] for r in range(nrows)] for c in range(ncols)]

random_mat = random_integer_matrix(3, 4)
transposed_mat = transpose_matrix(random_mat)

print("OG Matrix (3x4):")
for row in random_mat:
    print(row)

print("Transposed Matrix (4x3):")
for row in transposed_mat:
    print(row)
```

d. Create a function `matrix_multiply(left_matrix, right_matrix)` which performs the matrix multiplication. Do not assume the matrices follow a specific shape. Here's some tips and strategies:
* Determine how many rows and columns the left and right matrices have. Which numbers should match in order for us to have valid matrix multiplication?
* Iterating the elements of "list-of-lists" gives us the rows, but we need to iterate the columns for the right matrix.
* Use the `transpose_matrix` function, then iteratively apply `scalar_product`.

```python
def matrix_multiply(left_matrix, right_matrix):
    #flip right matrix, turning its coloumns to rows
    right_transposed = transpose_matrix(right_matrix)
    
    #get scalar product of each row in left matrix and each column in right matrix
    return [
        [scalar_product(left_row, right_col) for right_col in right_transposed]
        for left_row in left_matrix
    ]

A = random_integer_matrix(2, 3)
B = random_integer_matrix(3, 2)
C = matrix_multiply(A, B)

print("Matrix A (2x3):")
for row in A:
    print(row)
print("Matrix B (3x2):")
for row in B:
    print(row)
print("Result A x B (2x2):")
for row in C:
    print(row)
```

#### Problem 2 code summary

This script generates matrix functions using Python list comprehensions without external libraries (like NumPy). The matrix_multiply() function transposes the right matrix, turning its coloumns into rows so that the dot product can be computed using zip() and sum(). I wrote the code this way to follow idiomatic Python, making it clean and easy to read.
