### Problem 1: Hamiltonian Matrix and its Eigs

The time-independent Schrödinger equation is:

$$
\left[-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}+\hat{V}(x)\right]\psi(x)=\left[\hat{T} + \hat{V} \right]\psi=\mathscr{E}_n\psi(x).
$$

In quantum mechanics, we can represent operators (e.g. $\hat{T}$ and $\hat{V}$) as matrices, while the quantum state (in this case, $\psi$) is written as an $N \times 1$ vector

$$ \psi = 
\begin{bmatrix}
\psi_1 \\
\psi_2 \\
\vdots \\
\psi_i \\
\vdots \\
\psi_N
\end{bmatrix}
$$

where $\psi_i$ represents the value of the wave function at $x_i$. Now, we know that $\left[\hat{V}(x)\psi(x)\right]|_{x_i} \equiv V(x_i)\psi(x_i) = V_i \psi_i$. We can write this statement as:

$$\hat{V}
\begin{bmatrix}
\psi_1 \\
\psi_2 \\
\vdots \\
\psi_i \\
\vdots \\
\psi_N
\end{bmatrix} = 
\begin{bmatrix}
V_1\psi_1 \\
V_2\psi_2 \\
\vdots \\
V_i\psi_i \\
\vdots \\
V_N\psi_N
\end{bmatrix}
$$

This relation only holds true when $\hat{V}$ is an $N \times N$ square diagonal matrix of the form:

$$
\boxed{\hat{V}=
\begin{pmatrix}
V_1 & 0 & 0 & \cdots & 0 & 0 & 0 \\
0 & V_2 & 0 & \cdots & 0 & 0 & 0 \\
0 & 0 & V_3 & \cdots & 0 & 0 & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots & \vdots & \vdots \\
0 & 0 & 0 & \cdots & V_i & 0 & 0 \\
0 & 0 & 0 & \cdots & 0 & \ddots & 0 \\
0 & 0 & 0 & \cdots & 0 & 0 & V_N
\end{pmatrix}
}
$$

where $V_i$ denotes the potential's value at $x_i$. This is then what we call the "matrix representation" of the operator $\hat{V}$! At least, in the position basis -- more info on 142 or 241 :)

```python
import numpy as np
import matplotlib.pyplot as plt

def hamiltonian(x_pos, potential_array, mass, hbar=1.0):
    N = len(x_pos)
    dx = x_pos[1] - x_pos[0]

    diag = -2 * np.eye(N, k=0)
    off_diag_upper = 1 * np.eye(N, k=1)
    off_diag_lower = 1* np.eye(N, k=-1)

    # Setting up T based on the solution from paper work, using np.eye() 
    T = (- hbar**2 / ( 2 * mass * dx**2 )) * (diag + off_diag_upper + off_diag_lower)

    #Setting up V considering that it is a matrix for potential values
    V = np.diag(potential_array)

    return T + V
```

```python
import matplotlib.pyplot as plt
import numpy as np


def hamiltonian(x_pos, potential_array, mass, hbar=1.0):
    """Calculates the N x N Hamiltonian matrix H = T + V using central finite differences.

    Parameters
    ----------
    x_pos : numpy.ndarray
        1D array of spatial grid points.
    potential_array : numpy.ndarray
        1D array of potential values evaluated at x_pos.
    mass : float
        Particle mass m.
    hbar : float, optional
        Reduced Planck constant (default 1.0).

    Returns
    -------
    H : numpy.ndarray
        2D square matrix representing the Hamiltonian operator.
    """
    N = len(x_pos)
    dx = x_pos[1] - x_pos[0]

    # Kinetic energy matrix T using 3-point finite difference operator
    # -d^2/dx^2 f(x_i) approx -(f_{i+1} - 2*f_i + f_{i-1}) / (dx^2)
    diag = np.full(N, 2.0)
    off_diag = np.full(N - 1, -1.0)
    T = (
        (hbar**2)
        / (2.0 * mass * (dx**2))
        * (np.diag(diag) + np.diag(off_diag, k=1) + np.diag(off_diag, k=-1))
    )

    # Diagonal potential energy matrix V
    V = np.diag(potential_array)

    return T + V


# --- System Parameters ---
m = 1.0
hbar = 1.0
omega = 0.8
N = 1000
x_lim = 8.0

# Grid generation
x = np.linspace(-x_lim, x_lim, N)
dx = x[1] - x[0]

# Potential energy V(x) = 0.5 * m * omega^2 * x^2
V_harmonic = 0.5 * m * (omega**2) * (x**2)

# Compute Hamiltonian and solve eigenvalue problem
H = hamiltonian(x, V_harmonic, mass=m, hbar=hbar)
eigenvalues, eigenvectors = np.linalg.eigh(H)

# Normalize wavefunctions for continuous space: psi_cont = psi_disc / sqrt(dx)
eigenvectors_norm = eigenvectors / np.sqrt(dx)

# --- Plotting Problem 1 Report Figure ---
plt.figure(figsize=(9, 6), dpi=120)

# Plot harmonic potential well
plt.plot(x, V_harmonic, "k--", linewidth=1.5, label=r"$V(x) = \frac{1}{2}m\omega^2 x^2$")

# Plot first 5 wavefunctions offset by their eigenvalues
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd"]
for n in range(5):
    E_n = eigenvalues[n]
    psi_n = eigenvectors_norm[:, n]

    # Sign convention fix for clean display
    if psi_n[np.argmax(np.abs(psi_n))] < 0:
        psi_n = -psi_n

    # Scaling factor for visual clarity when overlaid on energy axes
    scale = 0.6
    plt.plot(x, E_n + scale * psi_n, color=colors[n], label=f"$n={n}$ ($E={E_n:.3f}$)")
    plt.axhline(E_n, color=colors[n], linestyle=":", alpha=0.5)

plt.xlim(-6, 6)
plt.ylim(-0.2, 4.2)
plt.xlabel("Position $x$ (a.u.)", fontsize=11)
plt.ylabel("Energy / Wavefunction Amplitude", fontsize=11)
plt.title("First 5 Stationary States of Harmonic Oscillator", fontsize=13, fontweight="bold")
plt.legend(loc="upper right", frameon=True)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig("problem1_harmonic_oscillator_wavefunctions.png", dpi=300)
plt.show()

# Print computed eigenenergies
print("First 5 Eigenvalues (Numerical vs Theoretical):")
for n in range(5):
    E_exact = hbar * omega * (n + 0.5)
    print(f"n={n}: Numerical = {eigenvalues[n]:.6f} | Theoretical = {E_exact:.6f}")
```

### Some codes for solving matrix problems

* solving matrix equations ($\bf{A}\vec{x} = \vec{b}$) -> `np.linalg.solve`
* LU-decomposition ($\bf{A} = \bf{L}\bf{U}$) -> `scipy.linalg.lu`
* matrix inversion ($\bf{A}^{-1}$) -> `np.linalg.inv`
* QR-decomposition ($\bf{A} = \bf{Q}\bf{R}$) -> `scipy.linalg.qr`
* eigenvalue problem ($\bf{A}\vec{x} = \lambda \vec{x}$) -> `np.linalg.eig`

Solving matrix problems by themselves isn't difficult. There is an abundance of linear algebra libraries that are ready to use: from the battle-tested ones (the classic [BLAS](https://www.netlib.org/blas/) and [LAPACK](https://www.netlib.org/lapack/)), HPC tools for large problems ([ScaLAPACK](https://netlib.org/scalapack/)), to hardware-accelerated ones often with GPUs ([cuBLAS](https://docs.nvidia.com/cuda/cublas/index.html) and [cuSolver](https://docs.nvidia.com/cuda/cusolver/index.html) for NVIDIA).
