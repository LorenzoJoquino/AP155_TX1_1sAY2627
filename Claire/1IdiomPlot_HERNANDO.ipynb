### Problem 1: 2D Wave interference (Gezerlis Problem 1.15)

Assume that the combined eﬀect of two objects dropped in a water basin can be modelled by:
$$ W(\textbf{r}) = A \sin (k|\textbf{r}-\textbf{r}_0|) + A  \sin (k|\textbf{r}-\textbf{r}_1|). $$ You can think of this as an updated version of Eq. (1.3) in the book, where we are dealing with the linear addition of sinusoidal waves, instead of electric-field contributions. For simplicity, take $A=1$, $k=2\pi/0.3$, $r_0 = (−1, 0)$ , and $r_1 = (1, 0)$. Since we are now faced with a density plot, you should employ, NOT `streamplot()` but, `imshow()` from `Matplotlib`; in each of $x$ and $y$ your plot should extend from $−4$ to $+4$.

```python
import matplotlib.pyplot as plt
import numpy as np

# Setting the given
A = 1
k = 2 * np.pi / 0.3
r0 = np.array([-1, 0])
r1 = np.array([1,0])

#Spatial grid definition
x = np.linspace(-4, 4, 4000) #Let's say it's 400
y = np.linspace(-4, 4, 4000)
X, Y = np.meshgrid(x,y)

#Solving the distance 
d0 = np.sqrt((X - r0[0]) ** 2 + (Y - r0[1]) ** 2)
d1 = np.sqrt((X - r1[0]) ** 2 + (Y - r1[1]) ** 2)

#Solving W(r)
W = A * (np.sin(k * d0) + np.sin(k * d1))

#Visualizing the density plot of W(r)
plt.figure(figsize=(7, 6))
im = plt.imshow(
    W, extent=[-4, 4, -4, 4], origin="lower", cmap="BuPu", aspect="equal"
)

plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.title("2D Wave Interference Pattern")

plt.savefig("2D_Wave_Interference_1.png", dpi=300, bbox_inches="tight") # This is how we save the image below under our working folder.
plt.show()
```

a. Create a function `random_integer_matrix(nrows, ncols)` that generates a matrix in the form of "list-of-lists" whose elements are random integers from $1$ to $9$. Each "sub-list" inside the main list is a row in the matrix.

```python
import random
def random_integer_matrix(nrows, ncols):
    matrix = []
    for i in range(nrows):
        row = []
        for i in range(ncols):
            row.append(random.randint(1,9))
        matrix.append(row)
    return matrix

random_integer_matrix(3, 4) # we call the function to create a 3x4 matrix
```

b. Create a function `scalar_product(list1, list2)` where you multiply each element at the same index in the two lists and sum/accumulate the results.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

sum (x * y for x, y in zip(list1, list2)) # This is shorter. zip pairs the elements into tupples.
```

```python
def scalar_product(list1, list2):
    total = 0 
    for i in range(len(list1)): # This accumulates products step-by-step
        total += list1[i] * list2[i]
    return total

#Letting values
list1 = [1, 2, 3]
list2 = [4, 5, 6]

print(scalar_product(list1, list2)) 
```

c. Create a function `transpose_matrix(orig_matrix)` which returns the transpose of any matrix also in "list-of-lists" form. Do not assume the size of the matrix; figure it out within the function. Double-check that your implementations above work. Create a $3\times4$ random integer matrix, print it out and its transpose. The values and shape between the two should be consistent.

```python
def transpose_matrix(orig_matrix):
    """Transposes any 2D list-of-lists matrix by dynamically finding its dimensions."""
    if not orig_matrix or not orig_matrix[0]:
        return []

    nrows = len(orig_matrix)
    ncols = len(orig_matrix[0])

    # Transposing matrix by swapping columns and rows where (r,c) becomes (c,r)
    return [[orig_matrix[r][c] for r in range(nrows)] for c in range(ncols)]


orig = random_integer_matrix(3, 4) # This is the original matrix from previous cell.
transposed = transpose_matrix(orig) # This is the transposed. We observe what happens from here.

print("Original Matrix (3x4):")
for row in orig:
    print(row)

print("\nTransposed Matrix")
for row in transposed:
    print(row)
    
```

d. Create a function `matrix_multiply(left_matrix, right_matrix)` which performs the matrix multiplication. Do not assume the matrices follow a specific shape. Here's some tips and strategies:
* Determine how many rows and columns the left and right matrices have. Which numbers should match in order for us to have valid matrix multiplication?
* Iterating the elements of "list-of-lists" gives us the rows, but we need to iterate the columns for the right matrix.
* Use the `transpose_matrix` function, then iteratively apply `scalar_product`.

```python
# This calculates the scalar product of two vectors
def scalar_product(list1, list2):
    return sum(x * y for x, y in zip(list1, list2)) 
    
# This builds the function of a transpose matrix
def transpose_matrix(orig_matrix):
    return [list(col) for col in zip(*orig_matrix)]

# This defines the function for matrix multiplication
def matrix_multiply(left_matrix, right_matrix):
    """Multiplies two matrices after verifying dimension compatibility."""
    left_rows = len(left_matrix)
    left_cols = len(left_matrix[0])
    right_rows = len(right_matrix)
    right_cols = len(right_matrix[0])

    # Validate matrix multiplication dimension rule: left columns == right rows
    if left_cols != right_rows:
        raise ValueError(
            f"Cannot multiply matrices: inner dimensions do not match "
            f"({left_rows}x{left_cols} and {right_rows}x{right_cols})."
        )

    # Transpose right_matrix so its columns become easily iteratable rows
    transposed_right = transpose_matrix(right_matrix)

    # Multiply each row of left_matrix by each row (original column) of transposed_right
    result = []
    for left_row in left_matrix:
        row_result = [
            scalar_product(left_row, right_col)
            for right_col in transposed_right
        ]
        result.append(row_result)

    return result

# Letting values
A = [[1, 2, 3], [4, 5, 6]]  # 2x3 matrix
B = [[7, 8], [9, 1], [2, 3]]  # 3x2 matrix

product = matrix_multiply(A, B)

print("Matrix A (2x3):")
for r in A:
    print(r)

print("\nMatrix B (3x2):")
for r in B:
    print(r)

print("\nProduct of Matrices A and B")
for r in product:
    print(r)
```
