### Report discussion

**What were the basis linear equations used to set up the matrix in each question of Problem 2?**

Part 1: Newton's second law on the knot with zero acceleration, resolved along $x$ and
$y$. Horizontally the two tensions cancel, $-T_1\cos\alpha + T_2\cos\beta = 0$;
vertically they carry the weight, $T_1\sin\alpha + T_2\sin\beta = Mg$. The 2 x 2 matrix
is invertible unless $\sin(\alpha + \beta) = 0$ (collinear ropes).

**Once you set up the matrices, what function did you use to solve the matrix equation? Summarize the mechanism behind the function.**

`np.linalg.solve` (Problem 1 at $N = 6$, Part 1) calls LAPACK `gesv`: LU factorisation
with partial pivoting, $PA = LU$, then $L\vec y = P\vec b$ by forward substitution and
$U\vec x = \vec y$ by back substitution. It costs about $\tfrac{2}{3}n^3$ flops and never
forms $A^{-1}$, which would be slower and less accurate.

`scipy.linalg.solve_banded` (Problem 1 at $N = 10^4$) calls LAPACK `gbsv`: the same
pivoted LU, but on the $(l + u + 1) \times N$ banded storage, so time and memory grow
linearly with $N$ for fixed bandwidth.

### Problem 1: (Circuit Analysis)


Consider a long chain of resistors wired up like this:

<img src="reschain.png" alt="Alt text" width=95%>

All the resistors have the same resistance~$R$.  The power rail at the top is at voltage~$V_+=5$V.  The problem is to find the voltages $V_1...V_N$ at the internal points in the circuit.

1. **Using Ohm's law and the Kirchhoff current law**, which says that the total net current flow out of (or into) any junction in a circuit must be zero, show that the voltages $V_1\ldots V_N$ satisfy the equations

$$3V_1 - V_2 - V_3 = V_+ $$
$$-V_1 + 4V_2 - V_3 - V_4 = V_+ $$
      
$$-V_{i-2} - V_{i-1} + 4V_i - V_{i+1} - V_{i+2} = 0 $$

$$-V_{N-3} - V_{N-2} + 4V_{N-1} - V_N = V_- = 0 $$
$$-V_{N-2} - V_{N-1} + 3V_N = V_- = 0 $$

Note: The assignment of $V_-$ is to make the equations symmetric

2. Express these equations in vector form~$A\vec{v} = \vec{w}$ and find the values of the matrix~$A$ and the vector~$\vec{w}$.

3. Write a program to solve for the values of the~$V_i$ when there are $N=6$ internal junctions with unknown voltages.  (Hint: All the values of $V_i$ should lie between zero and $5$V.  If they don't, something is wrong.)

4. Now repeat your calculation for the case where there are $N=10\,000$ internal junctions.  This part is not possible using standard tools like the *solve* function.  You need to make use of the fact that the matrix~$\vec{A}$ is banded.  

**Problem 1 notes**

*1. Equations.* Ohm's law gives the current leaving junction $i$ toward a neighbour at
$V_j$ as $(V_i - V_j)/R$; Kirchhoff's current law says these sum to zero at every
junction. Each internal junction is wired to the two junctions on either side, so for
$3 \le i \le N-2$
$$\frac{V_i - V_{i-2}}{R} + \frac{V_i - V_{i-1}}{R} + \frac{V_i - V_{i+1}}{R} + \frac{V_i - V_{i+2}}{R} = 0
\;\Rightarrow\; -V_{i-2} - V_{i-1} + 4V_i - V_{i+1} - V_{i+2} = 0 .$$
$R$ cancels, so the voltages do not depend on it. Junction 1 has three neighbours
($V_+$, $V_2$, $V_3$): $3V_1 - V_2 - V_3 = V_+$. Junction 2 has four ($V_+$, $V_1$, $V_3$,
$V_4$): $-V_1 + 4V_2 - V_3 - V_4 = V_+$. The last two junctions mirror these with
$V_- = 0$ in place of $V_+$.

*2. Matrix form.* $A\vec v = \vec w$ with $A$ pentadiagonal: 4 on the diagonal (3 in the
two corners), $-1$ on the first and second off-diagonals, 0 elsewhere;
$\vec w = (V_+, V_+, 0, \dots, 0)^T$.

*3, 4. Solving.* At $N = 6$ the dense $A$ is solved with `np.linalg.solve`. At
$N = 10^4$ a dense $A$ would take 800 MB and $O(N^3)$ work, so $A$ is stored in LAPACK
banded form (5 rows by $N$ columns) and solved with `scipy.linalg.solve_banded`, whose
cost is linear in $N$ for fixed bandwidth. The banded builder is checked against the
dense one at $N = 6$ before being used at $N = 10^4$.

```python
def resistor_chain_dense(N, V_plus=5.0):
    """Dense N x N matrix A and right-hand side w for the resistor chain.

    A is pentadiagonal: 4 on the diagonal (3 in the two corners) and -1 on
    the first and second off-diagonals. w holds V_plus in its first two
    entries and 0 elsewhere (V_minus = 0). Returns (A, w) as float arrays.
    """
    A = 4.0*np.eye(N)
    for k in (1, 2):
        A -= np.eye(N, k=k) + np.eye(N, k=-k)
    A[0, 0] = A[-1, -1] = 3.0
    w = np.zeros(N)
    w[:2] = V_plus
    return A, w


A6, w6 = resistor_chain_dense(6)
V6 = np.linalg.solve(A6, w6)
print("A (N = 6):\n", A6.astype(int))
print("w =", w6)
print("V =", V6.round(4))
print("all voltages within (0, 5) V:", np.all((V6 > 0) & (V6 < 5)))
```

```python
from scipy.linalg import solve_banded


def resistor_chain_banded(N, V_plus=5.0):
    """Same system as resistor_chain_dense, with A in LAPACK banded storage.

    Returns (ab, w). ab has shape (5, N): rows 0 and 1 hold the second and
    first superdiagonals, row 2 the diagonal, rows 3 and 4 the first and
    second subdiagonals, following ab[u + i - j, j] = A[i, j] with u = 2.
    """
    ab = np.zeros((5, N))
    ab[0, 2:] = ab[1, 1:] = -1.0      # superdiagonals
    ab[3, :-1] = ab[4, :-2] = -1.0    # subdiagonals
    ab[2, :] = 4.0
    ab[2, 0] = ab[2, -1] = 3.0
    w = np.zeros(N)
    w[:2] = V_plus
    return ab, w


ab6, _ = resistor_chain_banded(6)
print("banded == dense at N = 6:", np.allclose(solve_banded((2, 2), ab6, w6), V6))

N = 10_000
ab, w = resistor_chain_banded(N)
V = solve_banded((2, 2), ab, w)
print(f"N = {N}: first three V = {V[:3].round(4)}, last three V = {V[-3:].round(4)}")
print("all voltages within (0, 5) V:", np.all((V > 0) & (V < 5)))
```

**Problem 1 code summary**

The script turns Kirchhoff's current law at every junction into $A\vec v = \vec w$ and

expected answer, then a banded version stores only the five non-zero diagonals and

**Problem 2 notes**

*Part 1.* Rope 1 makes angle $\alpha$ and rope 2 angle $\beta$ with the ceiling, on
opposite sides of the mass. Force balance on the knot:
$$-T_1\cos\alpha + T_2\cos\beta = 0, \qquad T_1\sin\alpha + T_2\sin\beta = Mg,$$
so $A = \begin{pmatrix} -\cos\alpha & \cos\beta \\ \sin\alpha & \sin\beta \end{pmatrix}$
and $\vec b = (0,\, Mg)^T$. The function lives in `statics.py`, written by the cell below,
and is checked against the closed form $T_1 = Mg\cos\beta/\sin(\alpha+\beta)$,
$T_2 = Mg\cos\alpha/\sin(\alpha+\beta)$.

*Part 2* is omitted as instructed: its constraints are linearly dependent and mutually
inconsistent, so the masses cannot be determined.

```python
%%writefile statics.py

import numpy as np

G = 9.81  # gravitational acceleration, m/s^2


def rope_tensions(M, alpha_deg, beta_deg, g=G):
    """Tensions in two ropes holding a mass M (kg) at rest.

    Rope 1 makes angle alpha_deg and rope 2 angle beta_deg with the ceiling,
    in degrees, on opposite sides of the mass. Returns (T1, T2) in newtons
    by solving the force balance on the knot:
        -T1 cos(alpha) + T2 cos(beta) = 0
         T1 sin(alpha) + T2 sin(beta) = M g
    """
    a, b = np.radians(alpha_deg), np.radians(beta_deg)
    A = np.array([[-np.cos(a), np.cos(b)],
                  [ np.sin(a), np.sin(b)]])
    rhs = np.array([0.0, M*g])
    return tuple(np.linalg.solve(A, rhs))
```

```python
import statics

M, alpha, beta = 10.0, 30.0, 60.0            # kg, degrees, degrees
T1, T2 = statics.rope_tensions(M, alpha, beta)

# Hand computation: the 2 x 2 system solved by elimination.
a, b = np.radians(alpha), np.radians(beta)
T1_hand = M*statics.G*np.cos(b)/np.sin(a + b)
T2_hand = M*statics.G*np.cos(a)/np.sin(a + b)

print(f"T1 = {T1:.3f} N   (by hand: {T1_hand:.3f} N)")
print(f"T2 = {T2:.3f} N   (by hand: {T2_hand:.3f} N)")
print("computer matches hand computation:", np.allclose([T1, T2], [T1_hand, T2_hand]))
```

### Some codes for solving matrix problems

* solving matrix equations ($\bf{A}\vec{x} = \vec{b}$) -> `np.linalg.solve`
* LU-decomposition ($\bf{A} = \bf{L}\bf{U}$) -> `scipy.linalg.lu`
* matrix inversion ($\bf{A}^{-1}$) -> `np.linalg.inv`
* QR-decomposition ($\bf{A} = \bf{Q}\bf{R}$) -> `scipy.linalg.qr`
* eigenvalue problem ($\bf{A}\vec{x} = \lambda \vec{x}$) -> `np.linalg.eig`

Solving matrix problems by themselves isn't difficult. There is an abundance of linear algebra libraries that are ready to use: from the battle-tested ones (the classic [BLAS](https://www.netlib.org/blas/) and [LAPACK](https://www.netlib.org/lapack/)), HPC tools for large problems ([ScaLAPACK](https://netlib.org/scalapack/)), to hardware-accelerated ones often with GPUs ([cuBLAS](https://docs.nvidia.com/cuda/cublas/index.html) and [cuSolver](https://docs.nvidia.com/cuda/cusolver/index.html) for NVIDIA). 

The hardest part is actually composing the matrix. How do you transform a physical problem into a linear algebra problem?

* Circuit: https://personal.math.vt.edu/embree/cmda3606/chapter2.pdf
* Embree [main notes](https://personal.math.vt.edu/embree/cmda3606notes.pdf) + [lab manual](https://personal.math.vt.edu/embree/labman.pdf)
