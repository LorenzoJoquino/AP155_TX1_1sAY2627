### Report discussion

Increasing a Gaussian signal’s standard deviation ($\sigma$) broadens its time-domain waveform. Due to time-frequency uncertainty, this inversely affects the frequency domain, resulting in a narrower, more sharply peaked Fourier transform magnitude. Furthermore, lowering the sample size $N$ reduces the observation window, which degrades frequency resolution ($\Delta f = f_s / N$) and widens the spectral bins. Signal recovery remains possible up to a fundamental limit, that is, if $N$ drops too low, the broadened signal peaks irreversibly blend into the surrounding noise floor. Finally, Figure 1 illustrates the Nyquist-Shannon theorem by comparing sampling rates against a target signal frequency $f$. Using a non-aliased sampling frequency ($f_s > 2f$) accurately captures the original wave, whereas an aliased sampling frequency ($f_s < 2f$) artificially reconstructs it as a lower-frequency waveform due to undersampling.

### Problem 1: Exploring FT of varying waveforms

Given the waveforms from the pool below, produce a $1\times3$ plot containing the following: 
1) the signal, with its variations, 
2) the magnitude of its Fourier transform, and 
3) the phase information of the transform (up to you if unwrapped or not). 
    
Discussion: 2-3 sentences explaining how changing the shape of the input signal affects it transform. 
Ex: for the Gaussian waveform, how does it Fourier transform change based on increasing or decreasing sigma?

Waveforms you can try (prefarably non-sinusoidal):

```python
t = np.linspace(-10, 10, 200)
dt = t[1] - t[0]
sigma_list = [0.25, 0.5, 0.75, 1.0] #varying standard deviations
tolerance = 1e-7

fig, axes = plt.subplots(1, 3, figsize=(12, 6))

for sigma in sigma_list:
    y = np.exp(-(t**2) / (2 * sigma**2)) #original Gaussian signal
    
    #Fourier transform calculation using numpy (https://numpy.org/devdocs/reference/generated/numpy.fft.fftshift.html)
    Y = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(y))) #Fourier transformed signal
    freqs = np.fft.fftshift(np.fft.fftfreq(len(t), dt)) #shift from time domain to frequency domain
    
    magnitude = np.abs(Y)
    phase = np.angle(Y)
    phase[magnitude < tolerance] = 0 #remove phase noise when magnitude is zero

    axes[0].plot(t, y, label=rf'$\sigma = {sigma}$')
    axes[1].plot(freqs, magnitude, label=rf'$\sigma = {sigma}$')
    axes[2].plot(freqs, phase, label=rf'$\sigma = {sigma}$')

axes[0].set_title('Gaussian Signal')
axes[0].set_ylabel('Amplitude, y')
axes[0].set_xlabel('Time, t')
axes[0].set_xlim(-5, 5)

axes[1].set_title('Magnitude vs. Frequency Graph')
axes[1].set_ylabel('Magnitude, |Y|')
axes[1].set_xlabel('Frequency, f')
axes[1].set_xlim(-5, 5)

axes[2].set_title('Phase vs. Frequency Graph')
axes[2].set_ylabel('Phase, rad')
axes[2].set_xlabel('Frequency, f')
axes[2].set_xlim(-5, 5)
axes[2].set_ylim(-np.pi, np.pi)

for ax in axes:
    ax.legend()
    ax.grid(True)

plt.tight_layout()
plt.show()
```

######
As the standard deviation, $\sigma$, of the Gaussian signal increases, the waveform becomes broader and more spread out in the time domain. Consequently, its Fourier transform magnitude becomes narrower and more sharply peaked in the frequency domain.

### Fourier transform (from $t$-space to $\omega$-space)
$$ Y(\omega) = \int_{-\infty}^{+\infty} dt \frac{e^{-i\omega t}}{\sqrt{2\pi}}y(t)$$

### Inverse Fourier transform (from $\omega$-space to $t$-space)
$$ y(t) = \int_{-\infty}^{+\infty} d\omega Y(\omega) \frac{e^{i\omega t}}{\sqrt{2\pi}}$$

Something to think about: while we essentially do numerical integration, we do not impose h. Instead, we utilize the spacing of the dataset.

### Discrete FT/IFT

$$ Y_n = \frac{1}{\sqrt{2\pi}} \sum_{k=1}^N \left[(\cos(nk\theta)\Re\{y_k\}  +\sin(nk\theta)\Im\{y_k\} +i(\cos(nk\theta)\Im\{y_k\} −\sin(nk\theta)\Re\{y_k\})]\right]$$

$$ y_k = \frac{\sqrt{2\pi}}{N} \sum_{k=1}^N \left[(\cos(nk\theta)\Re\{Y_n\}  -\sin(nk\theta)\Im\{Y_n\} +i(\cos(nk\theta)\Im\{Y_n\} +\sin(nk\theta)\Re\{Y_n\})]\right]$$

Why bother it doing this way? 

It cleanly splits the real and imaginary part for both inputs and outputs, so that you can use a pair of floats rather than an aggregate complex data type (complex128 is just a pair of float64 anyway).

Using $Z=e^{−i\theta}, \rightarrow Z^{\pm nk} = e^{\mp ink\theta} = \cos(nk\theta)\mp i\sin(nk\theta)$.
