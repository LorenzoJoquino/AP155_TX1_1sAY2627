```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.interpolate as interp
from pathlib import Path

rng = np.random.default_rng(42)
```

### Problem 1: Interpolation of synthetic data

Generate random samples from a known sine function, then compare several interpolation methods. The interpolators only reconstruct values within the sampled domain; they should not be treated as reliable extrapolators outside it.

---

```python
'''
this is just another way to generate random samples

x = np.linspace(-4, 4, 100)
y = lambda x: np.sin(x)

x_sampled = np.random.choice(x, 50, replace = False)
y_sampled = y(x_sampled)
'''
```

```python
n = 100 #sample size

x_sampled = rng.uniform(-5, 5, n)
y_sampled = np.sin(x_sampled)

x = np.linspace(-5, 5, 200)
```

```python
sorted_indices = np.argsort(x_sampled)

x_sorted = x_sampled[sorted_indices]
y_sorted = y_sampled[sorted_indices]
```

```python
linear = np.interp(x, x_sorted, y_sorted) #linear interpolation

plt.figure(figsize=(9, 5))
plt.scatter(x_sampled, y_sampled, label = "Sampled data points", marker = ".")
plt.plot(x, np.sin(x), label = "True curve")
plt.plot(x, linear, label = "Linear interpolation curve")
plt.legend()
plt.show()
```

```python
cubic = interp.CubicSpline(x_sorted, y_sorted) #cubic interpolation

plt.figure(figsize=(9, 5))
plt.scatter(x_sampled, y_sampled, label = "Sampled data points", marker = ".")
plt.plot(x, np.sin(x), label = "True curve")
plt.plot(x, cubic(x), label = "Cubic interpolation curve")
plt.legend()
plt.show()
```

```python
x1 = np.linspace(-4.75, 4.75, 200)

nearest = interp.interp1d(x_sampled, y_sampled, kind = "nearest") #nearest interpolation

plt.figure(figsize=(9, 5))
plt.scatter(x_sampled, y_sampled, label = "Sampled data points", marker = ".")
plt.plot(x1, np.sin(x1), label = "True curve")
plt.plot(x1, nearest(x1), label = "Nearest interpolation curve")
plt.legend()
plt.show()
```

```python
quadratic = interp.make_interp_spline(x_sorted, y_sorted, k = 2) #quadratic interpolation

plt.figure(figsize=(9, 5))
plt.scatter(x_sampled, y_sampled, label = "Sampled data points", marker = ".")
plt.plot(x, np.sin(x), label = "True curve")
plt.plot(x, quadratic(x), label = "Quadratic interpolation curve")
plt.legend()
plt.show()
```
