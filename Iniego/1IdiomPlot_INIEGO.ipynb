### Problem 1: 2D Wave interference (Gezerlis Problem 1.15)

Assume that the combined eﬀect of two objects dropped in a water basin can be modelled by:
$$ W(\textbf{r}) = A \sin (k|\textbf{r}-\textbf{r}_0|) + A  \sin (k|\textbf{r}-\textbf{r}_1|). $$ You can think of this as an updated version of Eq. (1.3) in the book, where we are dealing with the linear addition of sinusoidal waves, instead of electric-field contributions. For simplicity, take $A=1$, $k=2\pi/0.3$, $r_0 = (−1, 0)$ , and $r_1 = (1, 0)$. Since we are now faced with a density plot, you should employ, NOT `streamplot()` but, `imshow()` from `Matplotlib`; in each of $x$ and $y$ your plot should extend from $−4$ to $+4$.

```python
import numpy as np
import matplotlib.pyplot as plt
    
def wave(r0,r1):
    """
        We define the wave and set parameters to it.
    """
    # We initialize a meshgrid
    x = np.linspace(-4,4,600)
    y = np.linspace(-4,4,600)
    X, Y = np.meshgrid(x,y)
    
    # Defining the wave parameters
    k = 2*np.pi/0.3
    A = 1
    # Setting the source positions
    
    # Distance from the sources
    dist_r0 = np.sqrt((X - r0[0])**2 + (Y - r0[1])**2)
    dist_r1 = np.sqrt((X - r1[0])**2 + (Y - r1[1])**2)

    # Initialize the sum of waves
    W1 = A*np.sin(k*abs(dist_r0))
    W2 = A*np.sin(k*abs(dist_r1))
    resultant = W1 + W2
    
    return resultant

def plottingWaves(resultant):
    """
        Parameter needed is a 1D array. We use this this to plot the waves.
        For more info on parameters, see imshow from matplotlib.
    """
    plt.imshow(resultant, extent=[-4,4,-4,4], origin='lower')
    plt.title("Sample Wave. ")
    plt.xlabel("X Axis")
    plt.ylabel("Y Axis")
    plt.grid(True)
    imageFilePath = "Figures/waveplot1.png"
    plt.savefig(imageFilePath)
    plt.show()
    
# Let us use four points to demonstrate
r0 = (-1,1)
r1 = (1,0)
r2 = (3,2)
r3 = (1,7)
resultant1 = wave(r0,r1)
resultant2 = wave(r2,r3)

# Modify the file name and location as needed
#plottingWaves(resultant1)
plottingWaves(resultant1)
```

Double-check that your implementations above work. Create a $3\times4$ random integer matrix, print it out and its transpose. The values and shape between the two should be consistent.

```python
def transpose_matrix(a):
    # Extract the rows and columns of a
    rows = len(a)
    cols = len(a[0])

    # Initialize a matrix with target dimensions based on original matrix
    b = [[0 for x in range(rows)] for x in range(cols)]
    for i in range(rows):
        for j in range(cols):
            b[j][i] = a[i][j]
    return b

a = [[3, 4, 1, 2], [9, 5, 6, 4]]
print(transpose_matrix(a))
```
