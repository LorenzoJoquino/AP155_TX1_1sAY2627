**1.** For Problem 2, the basis linear equations came from setting the net force on the hanging mass to be equal to zero (for both horizontal and vertical directions). Since the mass is static (stationary), the forces acting on it must be balanced.

So, for the horizontal direction,

$$
\sum F_x = 0
$$

which gives

$$
T_1\cos\alpha - T_2\cos\beta = 0.
$$

And for the vertical direction,

$$
\sum F_y = 0
$$

which gives us

$$
T_1\sin\alpha + T_2\sin\beta - Mg = 0.
$$

Here, $T_1$ and $T_2$ are the tensions in the two ropes, whereas, $Mg$ is the weight of the mass that is hanging. The sin and cos terms are used to separate the tension forces into horizontal and vertical components. These two equations contain the two unknown tensions. In the code, these equations were rearranged and written in matrix form in order to allow `np.linalg.solve()` to calculate the values of $T_1$ and $T_2$.

<br>

**2.** The function I used to solve the matrix equation was `np.linalg.solve()`. This function is used to solve a system of linear equations written in the form $A\mathbf{x}=\mathbf{b}$, wherein $A$ is the coefficient matrix, $\mathbf{x}$ is the vector that contains the unknown variables, and $\mathbf{b}$ is the vector that contains the known values. It works by using numerical linear algebra methods to solve the equations simultaneously and find the values of $\mathbf{x}$. In Problem 2, the unknown vector contained the two rope tensions, $T_1$ and $T_2$. The coefficient matrix was formed from the sin and cos terms of the rope angles, whereas, the right-hand-side vector contained the known force values. Hence, using `np.linalg.solve()` allowed the two tensions to be calculated directly without manually solving the equations.

Using Ohm's law,

$$
I=\frac{V_a-V_b}{R}.
$$

So at each junction, we can use Kirchhoff's law, (so total current is zero).

For $V_1$,

$$
\frac{V_1-V_+}{R}
+\frac{V_1-V_2}{R}
+\frac{V_1-V_3}{R}=0.
$$

$$
V_1-V_+ +V_1-V_2+V_1-V_3=0,
$$

$$
3V_1-V_2-V_3=V_+.
$$

For $V_2$,

$$
\frac{V_2-V_+}{R}
+\frac{V_2-V_1}{R}
+\frac{V_2-V_3}{R}
+\frac{V_2-V_4}{R}=0.
$$

$$
-V_1+4V_2-V_3-V_4=V_+.
$$

For the middle junctions, using the same process, it will give us

$$
-V_{i-2}-V_{i-1}+4V_i-V_{i+1}-V_{i+2}=0.
$$

For $V_{N-1}$,

$$
-V_{N-3}-V_{N-2}+4V_{N-1}-V_N=V_-=0.
$$

Lastly, for $V_N$,

$$
-V_{N-2}-V_{N-1}+3V_N=V_-=0.
$$

Hence, these are the equations satisfied by $V_1,\ldots,V_N$.

```python
# Banded matrix solver adapted from M. Newman, Computational Physics, 2nd edition. 

from numpy import copy

def banded(Aa, va, up, down):

    # Copy the inputs and determine the size of the system
    A = copy(Aa)
    v = copy(va)
    N = len(v)

    # Gaussian elimination
    for m in range(N):

        # Normalization factor
        div = A[up,m]

        # Update the vector first
        v[m] /= div
        for k in range(1,down+1):
            if m+k<N:
                v[m+k] -= A[up+k,m]*v[m]

        # Now normalize the pivot row of A and subtract from lower ones
        for i in range(up):
            j = m + up - i
            if j<N:
                A[i,j] /= div
                for k in range(1,down+1):
                    A[i+k,j] -= A[up+k,m]*A[i,j]

    # Backsubstitution
    for m in range(N-2,-1,-1):
        for i in range(up):
            j = m + up - i
            if j<N:
                v[m] -= A[i,j]*v[j]

    return v

# Number of internal junctions
N = 10000

# Banded matrix
A = np.zeros((5, N))

# Main diagonal
A[2, :] = 4
A[2, 0] = 3
A[2, -1] = 3

# First upper and lower diagonals
A[1, 1:] = -1
A[3, :-1] = -1

# Second upper and lower diagonals
A[0, 2:] = -1
A[4, :-2] = -1

# Vector w
w = np.zeros(N)
w[0] = 5
w[1] = 5

# Solve
v = banded(A, w, 2, 2)

print(v)
```

- If a pivot is placed $1.20$ [m] away from the right end, then the beam would balance if a $550$ [g] mass is placed on the left end of the beam.

```python
from math import sin, cos, radians, pi

def TensionCalculator(alpha, beta, M):

    # constant
    g = 9.81

    # Convert to radians
    alpha = radians(alpha)
    beta = radians(beta)

    # Coefficient Matrix
    u = np.array([[sin(alpha), sin(beta)], 
                  [-cos(alpha), cos(beta)]
                 ])
    # RHS Vector
    y = np.array([M*g, 0])

    # Calculation
    tensions = np.linalg.solve(u, y)

    return tensions
```

```python
# Trial
print(TensionCalculator(45, 45, 10))
```

### Some codes for solving matrix problems

* solving matrix equations ($\bf{A}\vec{x} = \vec{b}$) -> `np.linalg.solve`
* LU-decomposition ($\bf{A} = \bf{L}\bf{U}$) -> `scipy.linalg.lu`
* matrix inversion ($\bf{A}^{-1}$) -> `np.linalg.inv`
* QR-decomposition ($\bf{A} = \bf{Q}\bf{R}$) -> `scipy.linalg.qr`
* eigenvalue problem ($\bf{A}\vec{x} = \lambda \vec{x}$) -> `np.linalg.eig`

Solving matrix problems by themselves isn't difficult. There is an abundance of linear algebra libraries that are ready to use: from the battle-tested ones (the classic [BLAS](https://www.netlib.org/blas/) and [LAPACK](https://www.netlib.org/lapack/)), HPC tools for large problems ([ScaLAPACK](https://netlib.org/scalapack/)), to hardware-accelerated ones often with GPUs ([cuBLAS](https://docs.nvidia.com/cuda/cublas/index.html) and [cuSolver](https://docs.nvidia.com/cuda/cusolver/index.html) for NVIDIA). 

The hardest part is actually composing the matrix. How do you transform a physical problem into a linear algebra problem?
