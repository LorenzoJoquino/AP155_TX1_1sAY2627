The figure shows the intensity pattern produced when light from a point source (λ = 500 nm) passes through a circular aperture, plotted as I(r) = [J₁(kr)/kr]². The pattern is a bright central spot (the Airy disk) surrounded by concentric rings that get dimmer with distance. The dark rings occur where J₁(kr) = 0, at r ≈ 0.30, 0.56, and 0.81 μm, which sets the smallest detail a telescope can resolve.
```python
from math import pi, cos, sin, sqrt

def J1(m, x, N=1000):   # ito ay Bessel function J_m(x) via Simpson's rule 
    a, b = 0.0, pi
    h = (b - a) / N  # step size
    f = lambda t: cos(m * t - x * sin(t))  # integrand

    s = f(a) + f(b)
    for k in range(1, N, 2):   # odd points, weight 4
        s += 4 * f(a + k * h)
    for k in range(2, N, 2):   # even interior points, weight 2
        s += 2 * f(a + k * h)

    return (h / 3) * s / np.pi  # Simpson's rule result, normalized by pi

x_vals = np.linspace(0, 20, 1000)   # gawa x values from 0 to 20

for m in (0, 1, 2):  # Plot yung Bessel functions of orders 0, 1, and 2
    plt.plot(x_vals, [J1(m, x) for x in x_vals], label=f"$J_{m}(x)$")

plt.xlabel('x')
plt.ylabel('$J_m(x)$')
plt.title('Bessel Functions $J_0$, $J_1$, $J_2$')
plt.axhline(0, color='gray', lw=0.5)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
def J2(m, x, N=1000):  # ito ay Bessel via Simpson's rule
    h = np.pi/N  # step size
    x = np.array(x, dtype=float)
    f = lambda t: np.cos(m * t - x * np.sin(t))  # integrand
    
    s = f(0.0) + f(np.pi)  # endpoints, weight 1
    for k in range(1, N, 2):  # odd points, weight 4
        s += 4 * f(k * h)
    for k in range(2, N, 2):  # even interior points, weight 2
        s += 2 * f(k * h)
        
    return (h/3) * s / np.pi  # Simpson's rule (normalized by pi)

# Constants
wavelength = 0.5
k = 2*pi /wavelength
L = 1.0
n = 300

xs = np.linspace(-L, L, n)
X, Y = np.meshgrid(xs, xs)
r = np.sqrt(X**2 + Y**2)
kr = k * r

safe = np.where(kr == 0, 1.0, kr)  # iwas divide by zero kapag kr = 0
I = np.where(kr == 0, 0.25, (J2(1, safe) / safe) ** 2)  # I = (J1(kr)/kr)^2, at I(0) = 1/4 

# Plotting
plt.imshow(I, extent=[-L, L, -L, L], vmax=0.01, cmap='hot')
plt.xlabel('x (μm)')
plt.ylabel('y (μm)')
plt.title('Airy Disk Diffraction Pattern')
plt.colorbar(label='Relative Intensity')
plt.savefig('Diffraction12.png', dpi=300)
plt.show()
```
