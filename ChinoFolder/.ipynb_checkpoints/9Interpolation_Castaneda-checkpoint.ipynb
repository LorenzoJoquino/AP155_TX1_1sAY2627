## Section 2: Code

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import interpolate
from pathlib import Path
import urllib.request # we also put this

rng = np.random.default_rng(42)
```

### Problem 1: Interpolation of synthetic data

Generate random samples from a known sine function, then compare several interpolation methods. The interpolators only reconstruct values within the sampled domain; they should not be treated as reliable extrapolators outside it.

```python
# Set a random seed
np.random.seed(42)

x = np.sort(np.random.uniform(0, 2 * np.pi, 12)) # Generate 12 random x-values between 0 and 2π, then sort them for interpolation
y = np.sin(x) # Evaluate the sine function at each sampled x-value

x_dense = np.linspace(x.min(), x.max(), 500)  # make 500 evenly spaced x-values between the minimum & maximum of x
y_exact = np.sin(x_dense)  # find sin values from x_dense

# Interpolations
linear_interp = interpolate.interp1d(x, y, kind="linear")
cubic_interp = interpolate.interp1d(x, y, kind="cubic")
nearest_interp = interpolate.interp1d(x, y, kind="nearest")
quadratic_interp = interpolate.interp1d(x, y, kind="quadratic")

# use x_dense
y_linear = linear_interp(x_dense)
y_cubic = cubic_interp(x_dense)
y_nearest = nearest_interp(x_dense)
y_quadratic = quadratic_interp(x_dense)
```

```python
plt.figure(figsize=(10, 6))

# Exact sine function
plt.plot(x_dense, y_exact, "k--", label="Exact sine")

# Interpolation methods
plt.plot(x_dense, y_linear, label="Linear interpolation")
plt.plot(x_dense, y_cubic, label="Cubic interpolation")
plt.plot(x_dense, y_nearest, label="Nearest interpolation")
plt.plot(x_dense, y_quadratic, label="Quadratic  interpolation")

# Original data points
plt.scatter(x, y, color="red", zorder=5, label="Data points")

# Labels and formatting
plt.xlabel("x")
plt.ylabel("y")
plt.title("Comparison of Interpolation Methods")
plt.legend()
plt.grid(True)
plt.show()
```

### Problem 2: Interpolation of real-world data

Load the Dow Jones time-series dataset, randomly retain a specified fraction of observations, and compare the reconstructed curve with the original series. The sampling fractions are 50%, 30%, 10%, and 5%. 

```python
# Load dataset
url = "https://websites.umich.edu/~mejn/cp/data/dow.txt"
path = Path("dow.txt")

if not path.exists(): # checks whether the file does not exist
    urllib.request.urlretrieve(url, path)

dow = np.loadtxt(path)
```

```python
# Define orig time series
x = np.arange(len(dow))
fractions = [0.50, 0.30, 0.10, 0.05]
rng = np.random.default_rng(42)

# Plot orig series and interpolated reconstructions
fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes = axes.ravel()

for ax, fraction in zip(axes, fractions):
    # Randomly retain specified fraction of observations
    n_samples = max(2, int(fraction * len(dow)))
    indices = np.sort(rng.choice(len(dow), size=n_samples, replace=False))

    x_sample = x[indices]
    y_sample = dow[indices]

    # Linear interpolation
    interpolator = interpolate.interp1d(x_sample, y_sample, kind="linear", bounds_error=False, fill_value=np.nan)

    y_interp = interpolator(x)

    # Plot original series, reconstruction, and retained samples
    ax.plot(x, dow, label="Original series", alpha=0.7)
    ax.plot(x, y_interp, label="Linear interpolation")
    ax.scatter(x_sample, y_sample, s=12, label="Retained samples")

    ax.set_title(f"Sampling fraction: {fraction:.0%}")
    ax.set_xlabel("Time index")
    ax.set_ylabel("Dow Jones value")
    ax.legend()
    ax.grid(True)

plt.tight_layout()
plt.show()
```

### Problem 3: Compare interpolation methods on the sampled real-world data

Use a single sampling fraction to compare linear, quadratic, and cubic interpolation. Higher-order interpolation can produce smoother curves, but it can also overshoot between sparse or rapidly changing observations.

```python
# Sampling fraction
fraction = 0.10

# Sample observations
n_samples = max(4, int(fraction * len(dow)))
indices = np.sort(rng.choice(len(dow), size=n_samples, replace=False))

x_sample = x[indices]
y_sample = dow[indices]

# Create interpolation functions
linear_interp = interpolate.interp1d(x_sample, y_sample, kind="linear")
quadratic_interp = interpolate.interp1d(x_sample, y_sample, kind="quadratic")
cubic_interp = interpolate.interp1d(x_sample, y_sample, kind="cubic")

# Evaluate within the sampled domain
x_dense = np.linspace(x_sample.min(), x_sample.max(), 1000)

y_linear = linear_interp(x_dense)
y_quadratic = quadratic_interp(x_dense)
y_cubic = cubic_interp(x_dense)

# Plot
plt.figure(figsize=(12, 6))
plt.plot(x, dow, color="gray", alpha=0.5, label="Original series")
plt.scatter(x_sample, y_sample, color="black", s=20, label="Sampled observations", zorder=4)
plt.plot(x_dense, y_linear, label="Linear interpolation")
plt.plot(x_dense, y_quadratic, label="Quadratic interpolation")
plt.plot(x_dense, y_cubic, label="Cubic interpolation")
plt.xlabel("Time index")
plt.ylabel("Dow Jones value")
plt.title(f"Interpolation Method Comparison ({fraction:.0%} Sampling)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Problem  code summary

Problem 1: We drew 12 random x-values from 0 to 2π, computed their sine values, and built linear, cubic, and nearest-neighbor interpolators from them. And then, we evaluated all three on 500 evenly spaced points within the sampled range only (no extrapolation) and then plotted them against the exact sine.

Problem 2: We loaded the Dow Jones series and randomly kept 50%, 30%, 10%, and 5% of the observations. For each fraction, we rebuilt the curve with linear interpolation and plotted it against the original in a 2×2 grid, which showed the reconstruction getting rougher as fewer points were kept.

Problem 3: We fixed the sampling fraction at 10% and compared linear, quadratic, and cubic interpolation on the same sample, which we evaluated within the sampled range. The plot showed that quadratic and cubic are smoother but overshoot sharp turns, while linear stays within the range of its neighboring samples.
