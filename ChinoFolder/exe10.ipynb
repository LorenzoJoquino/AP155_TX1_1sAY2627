```python
import matplotlib.pyplot as plt
import numpy as np
import scipy
```

```python
N_sample = 200
N_unaliased = 1000
t = 10
dt = t/N_sample
dt_un = t/N_unaliased
t_sample = np.arange(N_sample)*dt
t_unaliased = np.arange(N_unaliased)*dt_un

cont_signal = lambda t,freq: np.cos(2*np.pi*freq*t)

# t_sample
freq = 1
discrete_signal = cont_signal(t_sample, freq)
dis_unaliased = cont_signal(t_unaliased, 18)

fig, ax = plt.subplots(figsize = (12,5))
ax.plot(t_sample, discrete_signal)

ax.plot(t_unaliased, dis_unaliased)
ax.set_xlim(0, 3)
fig.show()
```

```python
fft = np.fft.fft(discrete_signal)
fft_un = np.fft.fft(dis_unaliased)

plt.figure(figsize=(12, 5))
plt.plot(np.fft.fftshift(np.fft.fftfreq(N_sample, dt)),
         np.fft.fftshift(abs(fft)))
plt.plot(np.fft.fftshift(np.fft.fftfreq(N_unaliased, dt_un)),
         np.fft.fftshift(abs(fft_un)))

plt.xlim(-25, 25)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

N = 1000
t = np.linspace(-5, 5, N, endpoint=False)
dt = t[1] - t[0]

sigmas = [0.3, 0.7, 1.5]

fig, ax = plt.subplots(1, 3, figsize=(15, 4))

for sigma in sigmas:
    signal = np.exp(-t**2 / (2*sigma**2))

    # Fourier Transform
    fft = np.fft.fftshift(np.fft.fft(signal))
    freq = np.fft.fftshift(np.fft.fftfreq(N, dt))
    
    ax[0].plot(t, signal, label=f'sigma = {sigma}')
    ax[1].plot(freq, np.unwrap(abs(fft)*dt), label=f'sigma = {sigma}')
    ax[2].plot(freq, np.unwrap(np.angle(fft)), label=f'sigma = {sigma}')

ax[0].set_title('Gaussian Signals')
ax[0].set_xlabel('Time')
ax[0].set_ylabel('Amplitude')

ax[1].set_title('Magnitude of Fourier Transform')
ax[1].set_xlabel('Frequency')
ax[1].set_ylabel('Magnitude')
ax[1].set_xlim(-5, 5)

ax[2].set_title('Phase of Fourier Transform')
ax[2].set_xlabel('Frequency')
ax[2].set_ylabel('Phase (radians)')
ax[2].set_xlim(-5, 5)

for a in ax:
    a.legend()
    a.grid()

plt.tight_layout()
plt.show()
```
