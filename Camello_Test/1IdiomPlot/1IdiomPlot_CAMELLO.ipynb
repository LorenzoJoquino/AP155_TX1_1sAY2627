### Problem 1: 2D Wave interference (Gezerlis Problem 1.15)

Assume that the combined eﬀect of two objects dropped in a water basin can be modelled by:
$$ W(\textbf{r}) = A \sin (k|\textbf{r}-\textbf{r}_0|) + A  \sin (k|\textbf{r}-\textbf{r}_1|). $$ You can think of this as an updated version of Eq. (1.3) in the book, where we are dealing with the linear addition of sinusoidal waves, instead of electric-field contributions. For simplicity, take $A=1$, $k=2\pi/0.3$, $r_0 = (−1, 0)$ , and $r_1 = (1, 0)$. Since we are now faced with a density plot, you should employ, NOT `streamplot()` but, `imshow()` from `Matplotlib`; in each of $x$ and $y$ your plot should extend from $−4$ to $+4$.

```python
import matplotlib.pyplot as plt
import numpy as np

#Bounds of Graph
x = np.linspace(-4, 4, 100)
y = x.copy()
X, Y = np.meshgrid(x,y)   

#Given Variables
A = 1
k = (2*np.pi)/0.3
x0, y0 = -1, 0
x1, y1 = 1, 0

#Distance Calculation
dist0 = np.sqrt((X - x0)**2 + ((Y - y0)**2))
dist1 = np.sqrt((X - x1)**2 + ((Y - y1)**2))

#Defining Wave Interference Formula
WaveFunction = A*(np.sin(k*abs(dist0)) + np.sin(k*abs(dist1)))

#Plotting the Graph
plt.imshow(WaveFunction, cmap='Blues', extent = [-4, 4, -4, 4]) 
plt.xticks(range(-4, 5, 2))
plt.yticks(range(-4, 5))
plt.title("Wave Interference Plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.grid(True, alpha = 0.35)
plt.savefig("Wave Interference Plot", dpi=300)
```

d. Create a function `matrix_multiply(left_matrix, right_matrix)` which performs the matrix multiplication. Do not assume the matrices follow a specific shape. Here's some tips and strategies:
* Determine how many rows and columns the left and right matrices have. Which numbers should match in order for us to have valid matrix multiplication?
* Iterating the elements of "list-of-lists" gives us the rows, but we need to iterate the columns for the right matrix.
* Use the `transpose_matrix` function, then iteratively apply `scalar_product`.

```python
def scalar_product(list1, list2):
    total = 0
    for i in range(len(list1)): 
        product = list1[i]*list2[i]
        total += product
    return total

def transpose_matrix(orig_matrix):
    rows = len(orig_matrix)
    columns = len(orig_matrix[0])

    transposed_matrix = []
    for col in range(columns): 
        current_row = []
        for row in range(rows):
            current_row.append(orig_matrix[row][col])
        transposed_matrix.append(current_row)
    return transposed_matrix
#All functions above taken from previous sub-questions

def matrix_multiply(left_matrix, right_matrix):
    if len(left_matrix[0]) == len(right_matrix): #Simply checks whether the columns of left_matrix and rows of right_matrix are equal
        product_matrix = []
        transposed_right_matrix = transpose_matrix(right_matrix) #This allows the algorithm to compute the scalar products easier by transposing the right matrix and calculating by its rows instead of columns
        for left_row in left_matrix: #Repeats the following code for as many rows in the left matrix
            current_row = [] #The building block for creating each row of the resulting product matrix
            for right_col in transposed_right_matrix: #Repeats the following code for as many *columns* in the original right matrix
                dot_product = scalar_product(left_row, right_col) #Calculates the scalar product of the current values of the left matrix and transposed right matrix
                current_row.append(dot_product) #Appends the value to the current row 
            product_matrix.append(current_row) #Appends the entirety of the current row onto the main product matrix
        return product_matrix
    else:
        return print("Error: Matrices are incompatible for multiplication") #Returns this if the number of columns of the left matrix and the number of rows of the right matrix do not match
    """ 
    To explain simply, given the 2 matrices, it first checks whether the two are compatible for multiplication.
    If they are, a blank matrix (which gets appended with values throughout the loop) is created alongside the 
    transposing of the right matrix (as it is simpler to conduct scalar product with 2 rows than a row and a column). 
    Each row of the left matrix is then put through the scalar product function alongside the columns (which are now
    rows due to transposing) of the right matrix. The values for each of these row-column combinations are then saved
    and appended into the current row list. Once all columns have been calculated, the current row list, containing the
    values calculated, are then appended into the FIRST row of the product matrix. The code then moves on to the next row
    and repeats the same process until all rows have been completed. Through this code, multiplication between two 
    compatible matrices is made possible. If two matrices are incompatible, an error stating the two matrices are incompatible 
    is shown.
    """
left_matrix = [[1, 3], [5, 4]]
right_matrix = [[1, 3], [5, 4]]

print("The resulting product matrix is:", matrix_multiply(left_matrix, right_matrix))
```
